```python
import gradio as gr
import modal

# Modal class'ı çağır (yeni yöntem)
VetBERTClassifier = modal.Cls.from_name("vetbert-service", "VetBERTClassifier")

# Arayüzde kullanılacak fonksiyon
def diagnose(symptoms):
    classifier = VetBERTClassifier()
    result = classifier.predict.remote(symptoms)
    return "\n".join([f"{k}: {round(v * 100, 1)}%" for k, v in result.items()])

# Gradio arayüzü
gr.Interface(
    fn=diagnose,
    inputs=gr.Textbox(label="Kedinizin belirtilerini yazın"),
    outputs=gr.Textbox(label="Tahmini hastalık olasılıkları"),
    title="VetBERTDx Hastalık Tahmin Aracı"
).launch(share=True)
```
